低方差过滤法，去除方差低于阈值的数据

低方差的特征意味着该特征的所有样本值几乎相同，对预测影响极小，可以将其去掉1

```python
#生成 100 个服从标准正态分布（均值为 0，标准差为 1）的随机浮点数，返回一个长度为 100 的一维数组
a = np.random.randn(100)

print(a.shape)
#数组 a 的方差
print(np.var(a))
```

```python
# b = np.random.randn(100) * 0.1
#100 个服从正态分布（均值为 5，标准差为 0.1）的随机浮点数
b = np.random.normal(5, 0.1, size=100)
print(b.shape)
print(np.var(b))
```

```python
# 构建数据样本
#a 和 b 垂直堆叠成一个两行的二维数组，然后转置，成为一个形状为 (100, 2) 的二维数组
X = np.vstack((a, b)).T
print(X.shape)
```

```python
# 过滤掉方差低于指定阈值的特征
from sklearn.feature_selection import VarianceThreshold
```

```python
# 低方差过滤，0.01
#创建 VarianceThreshold 对象，设定方差阈值为 0.01
vt = VarianceThreshold(0.01)

#对 X 进行拟合与变换，移除方差小于等于 0.01 的特征，将保留的特征数据赋值给 X_new
X_new = vt.fit_transform(X)
print(X_new)
print(X_new.shape)
```
