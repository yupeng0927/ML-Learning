```python
import pandas as pd
```

```python
# 读取数据
advertising = pd.read_csv('../../data/advertising.csv')
#打印该 DataFrame 的前 5 行数据
print(advertising.head())
# 输出 DataFrame advertising 中各数值列的统计摘要，
# 包括：计数、均值、标准差、最小值、25% 分位数、中位数（50%）、75% 分位数和最大值。
print(advertising.describe())
print(advertising.shape)
```

```python
# 去掉第一列，inplace=True：在原 DataFrame 上修改
advertising.drop(advertising.columns[0], axis=1, inplace=True)
print(advertising.shape)
print(advertising.head())
```

```python
# 去掉空值
advertising.dropna(inplace=True)
```

```python
# 构建 X 和 y
X = advertising.drop('Sales',axis=1)  # X是三个特征构成的矩阵
```

```python
y = advertising["Sales"]
```

```python
# 计算pearson相关系数,
# 计算 X 中每一列与序列 y 之间的皮尔逊相关系数，并打印结果
print(X.corrwith(y, method="pearson"))
```

```python
# 相关系数矩阵
corr_matrix = advertising.corr(method="pearson")
print(corr_matrix)
```

```python
import matplotlib.pyplot as plt
# ＃将材关系数短阵面成热力图
# 用于绘制统计图形
import seaborn as sns
```

```python
# annot=True ：显示数值。cmap="coolwarm"：颜色方案。fmt='.2f'：两位小数
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt='.2f')
plt.title("Pearson Correlation")
plt.show()
```
