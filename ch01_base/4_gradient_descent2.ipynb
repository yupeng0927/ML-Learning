```python
 # 构造目标函数 损失函数
def J(x):
    return (x ** 2 - 2) ** 2
# 梯度函数 导函数
def gradient(x):
    return 4 * x ** 3 - 8 * x
```

```python
# 用列表保存点的变化轨迹
x_list = []
y_list = []
```

```python
# 1. 初始化参数（x）和超参数
x = 1
alpha = 0.02
```

```python
import numpy as np

# 4. 重复迭代，当梯度接近于0的时候停止
while np.abs(grad:= gradient(x)) > 1e-6:
    # x_list.append(x)
    # y_list.append(grad)
# while np.abs(grad := gradient(x)) > 1e-10:
    # 记录当前点坐标
    x_list.append(x)
    y_list.append(J(x))
    print(f"x: {x}, \tf(x) = {J(x)}")
    
    # 2. 计算梯度

    # 3. 更新参数
    x = x - alpha * grad
print(len(x_list))
```

```python
# 画图
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(0.9, 1.6, 0.01)
plt.plot(x, J(x))   # 绘制函数曲线
plt.plot(x_list, y_list, 'r')
plt.scatter(x_list, y_list, color='r')
plt.show()
```

```python
# 局部放大对比
# 1 行 2 列、共 2 个子图​ 的图形窗口，画布尺寸为 15×4 英寸
fig, ax = plt.subplots(1, 2, figsize=(15, 4))
# -2 到 2 之间（不包含 2）以 0.01 为步长生成一组等差数值
x = np.arange(-2, 2, 0.01)
ax[0].plot(x, J(x))   # 绘制函数曲线
ax[0].plot(x_list, y_list, 'r')
ax[0].scatter(x_list, y_list, color='r')

x2 = np.arange(1.399, 1.425, 0.001)
x_list2 = x_list[1:]
y_list2 = y_list[1:]
ax[1].plot(x2, J(x2))   # 绘制函数曲线
ax[1].plot(x_list2, y_list2, 'r')
ax[1].scatter(x_list2, y_list2, color='r')

plt.show()
```
