```python
# 定义目标函数（模拟损失函数）
def f(x):
    return x ** 2
# 定义导函数（计算梯度的函数）
def gradient(x):
    return 2 * x
```

```python
# 用列表保存点的变化轨迹
x_list = []
y_list = []
```

```python
# 1. 初始化参数（x）和超参数
x = 1
alpha = 0.1
```

```python
# 4. 重复迭代 100 次
for i in range(100):
    # 记录当前点坐标
    x_list.append(x)
    y_list.append(f(x))
    print(f"x: {x}, \tf(x) = {f(x)}")
    
    # 2. 计算梯度
    grad = gradient(x)
    
    # 3. 更新参数
    x = x - alpha * grad
```

```python
# 画图
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(-1, 1, 0.01)
plt.plot(x, f(x))   # 绘制函数曲线
plt.plot(x_list, y_list, 'r')
plt.scatter(x_list, y_list, color='r')
plt.show()
```
