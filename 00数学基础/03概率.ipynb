```python
import numpy as np
```

```python
# 随机生成（0，1）均匀分布的数
a=np.random.rand(5)
print(a)
```

```python
b=np.random.rand(2,3)
print(b)
```

```python
# 随机生成（0，100）2行3列的矩阵
c=np.random.randint(0,100,size=(2,3))
print(c)
```

```python
#随机生成服从标准正态分布的数  均值为0，标准差为1，方差为1
d=np.random.randn(2,3)
print(d)
print((d+3)*10)
```
