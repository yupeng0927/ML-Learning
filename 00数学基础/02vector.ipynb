```python
# 相加
z=np.array([7,8,9])
print(z+x)
(z+x).shape
```

```python
# 计算范数
lo_norm=np.linalg.norm(x,ord=0) #L 0 范式
print(lo_norm)

l2_norm=np.linalg.norm(x,ord=2) #L 2范式
print(l2_norm)
```
