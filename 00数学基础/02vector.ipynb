```python
import numpy as np
from jinja2.ext import do
from numpy.ma.core import dot
```

```python
# 定义向量
x=np.array([1,2,3])
print(x)
x.shape
```

```python
# 向量转制
x.T
```

```python
# 定义二维
y=np.array([[1,2,3]])
print(y)
y.shape
```

```python
# 相加
z=np.array([7,8,9])
print(z+x)
(z+x).shape
```

```python
# 向量相乘  对应位置相互乘
print(z * x)
```

```python
# 点乘
print(x.dot(z))
print(np.dot(x,z))
print(x @ z)
```

```python
# 计算范数
lo_norm=np.linalg.norm(x,ord=0) #L 0 范式
print(lo_norm)

l2_norm=np.linalg.norm(x,ord=2) #L 2范式
print(l2_norm)
```

```python
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print((a.shape))
print((a * b))
print(np.dot([1,2],[3,4]))
print(a@b)
```
